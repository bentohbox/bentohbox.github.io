# Portfolio Optimization using Markowitz Theory

## **Project Overview**
This project demonstrates the application of Modern Portfolio Theory (MPT) to optimise a portfolio of stocks. Using historical stock price data, it calculates the expected returns, risks, and Sharpe ratios of various portfolios, ultimately identifying an optimal portfolio with the maximum Sharpe ratio.

## **Key Objectives**
1. **Data Collection**: Fetch historical stock prices for a set of selected assets over the past two years. Why 2 years only? Using data that stretches too far into history might reduce recency and statistical reliability.
2. **Data Preprocessing**: Clean and normalise the data to ensure consistency and comparability. Why? Stock prices vary largely and there is a need to ensure they are comparable in scale.
3. **Log Returns Calculation**: Calculate daily logarithmic returns to understand historical price movements.
4. **Portfolio Simulation**: Generate random portfolios with varying weights to estimate the risk-return characteristics of different combinations.
5. **Optimisation**: Use numerical optimisation techniques to maximize the Sharpe ratio, identifying the optimal allocation of assets.
6. **Visualisation**: Display portfolio distributions and the optimal portfolio on a risk-return scatter plot for better interpretability.

## **Data Sources**
- Historical stock prices were retrieved using the `yfinance` library:
  - Amazon (AMZN)
  - Palo Alto Networks (PANW)
  - JPMorgan Chase (JPM)
  - Texas Instruments (TXN)

  - Vital Energy (VTLE)

## **Project Workflow**
1. **Data Downloading**: 
   - Collected daily closing prices for the selected stocks from Yahoo Finance
   - Addressed missing data through forward and backward filling to ensure completeness

2. **Log Return Computation**:
   - Normalized historical prices and calculated daily log returns to measure relative price changes effectively

3. **Portfolio Simulation**:
   - Simulated 10,000 random portfolios, calculating expected returns and volatilities for each
   - Evaluated the Sharpe ratio (return-to-risk ratio) for each portfolio

4. **Optimisation**:
   - Formulated and solved a constrained optimization problem to maximize the Sharpe ratio
   - Constraints included ensuring portfolio weights sum to 1 and remain within [0, 1]

5. **Visualisation**:
   - Created scatter plots to visualize portfolios' risk-return characteristics
   - Highlighted the optimal portfolio with its Sharpe ratio, expected return, and volatility

## **Key Results**
- **Optimal Portfolio**: Achieved through numerical optimization using the Sequential Least Squares Programming (SLSQP) method
- **Sharpe Ratio**: Provided insights into the risk-adjusted performance of the portfolio
- **Allocation**: Displayed optimal weights for the selected stocks, showing how to allocate capital efficiently
